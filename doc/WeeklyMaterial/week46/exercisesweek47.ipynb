# Exercises week 47 (the last weekly exercise): your proposal for Project 3

**FYS-STK3155/4155, fall 2026 — published in week 46; deadline Friday November 20 (midnight), one submission per group**

There is no coding exercise this week. The exercise is the **proposal for
Project 3** that your group starts drafting in the Tuesday session of week
46: a short slide deck of **3–5 slides** that says what you want to do, with
which data and methods, and how you plan to get there before the deadline on
December 11. You present it — **3–5 slides, 3–5 minutes per group** — in the
final flipped classroom on Tuesday November 17, you get our reaction and that
of the other groups, and you hand in the revised version by **Friday November
20**. This is the last chance to adjust the scope before you commit to it, so
make the proposal concrete enough that we can tell you whether it is too
much, too little or just right.

Read the full project text first (`Project3.pdf`): the three paths, the
requirements common to all of them, the summary of requirements for the path
you choose, and the updated rules on the use of LLMs.

## What the proposal must contain (roughly one slide per item, 3–5 slides in all)

1. **The question and the data.** One sentence stating what you want to find
   out. The data: what they are, where they come from (link), how many
   samples and features, what the target is, and one or two references to
   earlier analyses.
   *Path A:* which equation, with which initial and boundary conditions,
   and what the analytical or trusted reference solution is.
   *Path C:* which synthetic setup — the function and noise level, the
   Gaussian mixture — and which quantities are known exactly (noise
   variance, Bayes error, closed-form bias and variance).
2. **Methods and baseline.** Which two or three methods you will use and
   which one is the baseline; what is your own code (from projects 1 and 2
   or new) and what comes from a library; which quality measures you will
   report; how the hyperparameters will be chosen (cross-validation, a
   validation split, the out-of-bag error) and how the test data are kept
   apart.
3. **Plan and risks.** A timeline from now to December 11 with the division
   of labour in the group; the computing you need and where you will run it;
   the one thing most likely to go wrong and what you will do if it does.
4. **(Optional) The figures you expect.** The two or three key figures of
   the report, sketched before you have results: what is on the axes, and
   what the curves should look like if your expectation is right.

## If you propose a data set or a problem of your own

Welcome — include the following on page 1 or an extra page:

- where the data can be downloaded (without registration if possible), in
  which format, and under which licence;
- the size: number of samples, features, classes or targets, class balance;
- whether a canonical train/test split exists, or how you will make one;
- at least one published analysis, benchmark table or documented
  leaderboard to compare with;
- why the methods of the course fit this problem (tabular data → trees,
  ensembles, logistic regression, networks; images → convolutional networks;
  sequences → recurrent networks; a known equation → path A).

If you are unsure whether a data set or a problem is suitable, write to us
before Tuesday rather than after.

## How we will use it

Every group presents its proposal as it stands on Tuesday November 17 —
3–5 slides, 3–5 minutes; keep the slides readable from the back of the room
(few words, one message per slide). We comment on the scope, the baseline,
the comparison with the literature and the risks; you revise the proposal
with that feedback and hand in the final version by Friday November 20, and
you keep it as the first page of your project plan. The proposal is not
graded, but it is the last weekly exercise and it counts as such.
