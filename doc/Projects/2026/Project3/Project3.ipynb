5.  **Averaging removes variance, not bias.** Bagging and random forests improve on a single tree because a tree is unstable; averaging a biased model leaves the bias where it was. Boosting, and more expressive models, are the routes that attack the bias.

6.  **Verification comes before results.** A finite-difference solution against the analytical one, a network against Scikit-Learn or PyTorch, a measured variance against a formula, a gradient against automatic differentiation: the tests that make an implementation trustworthy belong in the report.

7.  **Reproducibility is part of the science.** Seeds, the hyperparameters you settled on, a `README` file that lets us rerun your notebooks, and citations of the data sets, the software and the literature you used.

### Path A, part a): Setting up the problem

The physical problem can be that of the temperature in a rod of length $L=1$, held at zero temperature at both ends. We are looking at the one-dimensional problem

$$
\frac{\partial^2 u(x,t)}{\partial x^2} = \frac{\partial u(x,t)}{\partial t},
  \qquad t>0,\quad x\in[0,L],
$$

or

$$
u_{xx} = u_t ,
$$

with the initial condition, i.e. the condition at $t=0$,

$$
u(x,0) = \sin(\pi x), \qquad 0<x<L,
$$

with $L=1$ the length of the $x$-region of interest, and the boundary conditions

$$
u(0,t) = 0, \qquad u(L,t) = 0, \qquad t\ge 0 .
$$

The function $u(x,t)$ can be the temperature of the rod. As time increases the initial profile decays towards the stationary state $u=0$.

We will limit ourselves to the so-called explicit forward Euler algorithm, with a forward formula for the time derivative and a centered difference in space,

$$
u_t \approx \frac{u(x,t+\Delta t)-u(x,t)}{\Delta t}
      = \frac{u(x_i,t_j+\Delta t)-u(x_i,t_j)}{\Delta t},
$$

and

$$
u_{xx} \approx \frac{u(x+\Delta x,t)-2u(x,t)+u(x-\Delta x,t)}{\Delta x^2}
       = \frac{u(x_i+\Delta x,t_j)-2u(x_i,t_j)+u(x_i-\Delta x,t_j)}{\Delta x^2}.
$$

## Path C: The statistics and mathematics behind the methods

This path is for those of you who want to understand *why* the methods of the course behave as they do, rather than to apply them to a new data set. You will work mostly with synthetic data, where the true function, the noise and the Bayes error are known exactly, so that every quantity we have talked about during the semester – bias, variance, the expected test error, the correlation between the members of an ensemble, the bias of an error estimator – can be *measured* and compared with what the mathematics predicts. Derivations and controlled numerical experiments carry equal weight here; the report should read as a small study in statistical learning, with the theory in the methods section and the experiments confirming (or refining) it in the results.

The data we propose are

- for regression, a known function with Gaussian noise of known variance $\sigma^2$, for example the one-dimensional Runge function $f(x)=1/(1+25x^2)$ of projects 1 and 2, or the two-dimensional Runge function or Franke’s function; you draw as many independent training sets of size $n$ as you need, and a large independent test set;

- for classification, a mixture of two (or more) Gaussian distributions in $p$ dimensions with known means and covariances. For two classes with a common covariance matrix $\boldsymbol{\Sigma}$ and means $\boldsymbol{\mu}_0$, $\boldsymbol{\mu}_1$, the Bayes classifier is linear and the Bayes error is $\Phi(-\Delta/2)$, with $\Delta^2=(\boldsymbol{\mu}_1-\boldsymbol{\mu}_0)^T\boldsymbol{\Sigma}^{-1}(\boldsymbol{\mu}_1-\boldsymbol{\mu}_0)$ the Mahalanobis distance and $\Phi$ the standard normal cumulative distribution; with different covariances or with several components per class the Bayes boundary is curved and the Bayes error can be computed numerically from the known densities.

You may add one real data set at the end to see how much of what you found carries over, but the heart of this path is the controlled experiments.

```python
# Path C, part a) is analytical and belongs in the theory section of your report.
# A starting point for the data of parts b)-e): a two-class Gaussian mixture
# with a known Bayes error.
import numpy as np
from scipy.stats import norm

rng = np.random.default_rng(2026)      # fix the seed for reproducible results

def gaussian_mixture(n, mu0, mu1, Sigma, prior1=0.5, rng=rng):
    """Two classes with a common covariance: n samples, labels 0/1."""
    y = (rng.random(n) < prior1).astype(int)
    X = np.where(y[:, None] == 1,
                 rng.multivariate_normal(mu1, Sigma, n),
                 rng.multivariate_normal(mu0, Sigma, n))
    return X, y

def bayes_error(mu0, mu1, Sigma):
    """Equal priors, common covariance: Phi(-Delta/2) with Delta the Mahalanobis distance."""
    d = np.asarray(mu1) - np.asarray(mu0)
    Delta = np.sqrt(d @ np.linalg.solve(Sigma, d))
    return norm.cdf(-Delta / 2)

mu0, mu1 = np.array([0.0, 0.0]), np.array([2.0, 1.0])
Sigma = np.array([[1.0, 0.5], [0.5, 1.5]])
X, y = gaussian_mixture(1000, mu0, mu1, Sigma)
print("Bayes error for these parameters:", round(bayes_error(mu0, mu1, Sigma), 4))
print("class fractions in the sample:", np.bincount(y) / len(y))
```
